#OPERADORES

Adição (+)
Subtração (-)
Multiplicação (*)
Divisão (/)
Divisão interia (//)
Módulo (%)
Potência (**)



```python
#Funcionam como na matemática.
#Se os dois números forem inteiros (int),
#o resultado é inteiro.
#Se pelo menos um for decimal (float),
#o resultado é decimal.

print(10+5)
print(10-15)
print(4*2.5)
```

```python
#Atenção

print("oi"+"!")#concatenar
print("ha"*3)#repetição
print('5'+3)#não pode somar texto com número
```

```python
#DIVISÃO: / e //

#A divisão / sempre devolve um float,
#mesmo quando a conta é exata. Já a divisão inteira //
#descarta a parte decimal, arredondando para baixo.

print(10/2)
print(10/3)
print(10//3)
```

```python
#ATENÇÃO

#Dividir por zero gera erro

print(10//0)
```

```python
#MÓDULO

#O operador % devolve o resto da divisão.
#Ele é muito usado para:

#.Saber se um número é par ou ímpar
#.Saber se um número é divisível por outro

print(20 % 10)
print(100 % 2)
```

```python
#POTÊNCIA

print(2**3)#2*2*2
```

```python
#ORDEM DE PRECEDÊNCIA

#Quando há vários operadores na mesma expressão,
#o Python segue uma ordem, igual à matemática:

'''
Prioridade

()
**
*, / , // , %
+, -

'''
```

```python
print(2 + 3 * 4)
print((2 +  3) * 4)
print(10 - 4 / 2)
print(2 ** 3 ** 2)#feita da direita p/esquerda
print(-2 ** 2)
print(2 + 2 ** 2 ** 4)
```

#ATIVIDADE

```python
#Ler dois números e mostrar todas as operações

a = int(input('Digite o 1º número: '))
b = int(input('Digite o 2º número: '))

soma = a + b
subtracao = a - b
multi = a * b
divisao = a / b
divisao_int = a // b
modulo = a % b
potencia = a ** b

print(f'Soma: {soma}')
print(f'Divisão: {divisao}')
print(f'Divisão Interia: {divisao_int}')
print(f'Multiplicação: {multi}')
print(f'Módulo: {modulo}')
print(f'Potência: {potencia}')
```

```python
#MÉDIA DE NOTAS

'''
Ler três notas de um aluno e mostrar a média
'''

n1 = float(input('Nota 1:'))
n2 = float(input('Nota 2:'))
n3 = float(input('Nota 3:'))

media = (n1 + n2 + n3) / 3

print(f'A média é: {media}')
```

```python
#Calcular IMC

peso = float(input('Digite seu peso: '))
altura = float(input('Digite sua altura: '))

imc = peso / altura ** 2

print(f'Seu IMC é: {imc}')
```
